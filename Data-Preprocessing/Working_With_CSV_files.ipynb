```python
def rename(name):
    if name == 'Royal Challengers Bangalore':
        return 'RCB'
    else:
        return name
```

```python
rename('Royal Challengers Bangalore')
```

```python
# Convertors
pd.read_csv('Downloads/IPL Matches 2008-2020.csv',converters={'team1':rename})
```

```python
def rename(name):
    if name == 'Mumbai Indians':
        return 'MI'
    else:
        return name
```

```python
pd.read_csv('Downloads/IPL Matches 2008-2020.csv',converters={'team2':rename})
```
